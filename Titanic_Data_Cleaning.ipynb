## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
```

## 2. Load Titanic Dataset

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully")
print(df.head())
```

## 3. Explore Dataset

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())
```

## 4. Check Missing Values

```python
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
```

## 5. Handle Missing Values

```python
# Fill missing Age values with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing Embarked values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop Cabin column due to many missing values
df = df.drop(columns=['Cabin'])

print("Missing values handled successfully")
```

## 6. Remove Duplicate Records

```python
print("Duplicate records before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after cleaning:", df.duplicated().sum())
```

## 7. Rename Column

```python
df.rename(columns={'Sex': 'Gender'}, inplace=True)

print("Column renamed successfully")
print(df.columns.tolist())
```

## 8. Verify Cleaned Dataset

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nCleaned Dataset Shape:", df.shape)

print("\nFirst Five Rows of Cleaned Dataset:")
print(df.head())
```

## 9. Save Cleaned Dataset

```python
df.to_csv("cleaned_titanic.csv", index=False)

print("Dataset cleaned and saved successfully!")
```
